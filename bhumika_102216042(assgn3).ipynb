```python
#ASSIGNMENT 3 :
#exercise 1 :

import threading
import time

def print_numbers(thread_name):
    for i in range(1, 11):
        print(f"{thread_name}: {i}")
        time.sleep(1)
thread1 = threading.Thread(target=print_numbers, args=("Thread 1",))
thread2 = threading.Thread(target=print_numbers, args=("Thread 2",))

#starting both threads
thread1.start()
thread2.start()

#waiting for both the threads to finish
thread1.join()
thread2.join()

print("execution completed")
```

```python

import os
import shutil

class LocalCloudStorage:
    def __init__(self, storage_dir):
        """
        Initialize the local cloud storage with a specified directory.
        :param storage_dir: The directory to act as cloud storage.
        """
        self.storage_dir = storage_dir
        if not os.path.exists(self.storage_dir):
            os.makedirs(self.storage_dir)

    def upload_file(self, file_path, cloud_path):
        """
        Simulate uploading a file to the cloud storage.
        :param file_path: The path to the local file to upload.
        :param cloud_path: The path in the cloud storage to save the file.
        """
        full_path = os.path.join(self.storage_dir, cloud_path)
        os.makedirs(os.path.dirname(full_path), exist_ok=True)
        shutil.copy2(file_path, full_path)
        print(f"File uploaded to {full_path}")

    def download_file(self, cloud_path, local_path):
        """
        Simulate downloading a file from the cloud storage.
        :param cloud_path: The path in the cloud storage to download the file from.
        :param local_path: The path to save the downloaded file locally.
        """
        full_path = os.path.join(self.storage_dir, cloud_path)
        shutil.copy2(full_path, local_path)
        print(f"File downloaded from {full_path} to {local_path}")

    def list_files(self, cloud_dir=""):
        """
        List all files in a specific directory in the cloud storage.
        :param cloud_dir: The cloud directory to list the files from.
        """
        full_path = os.path.join(self.storage_dir, cloud_dir)
        if not os.path.exists(full_path):
            print(f"No such directory: {full_path}")
            return []

        files = []
        for root, _, filenames in os.walk(full_path):
            for filename in filenames:
                files.append(os.path.relpath(os.path.join(root, filename), self.storage_dir))

        return files

    def delete_file(self, cloud_path):
        """
        Simulate deleting a file from the cloud storage.
        :param cloud_path: The path in the cloud storage to delete the file.
        """
        full_path = os.path.join(self.storage_dir, cloud_path)
        if os.path.exists(full_path):
            os.remove(full_path)
            print(f"File {full_path} deleted.")
        else:
            print(f"File {full_path} does not exist.")

```
