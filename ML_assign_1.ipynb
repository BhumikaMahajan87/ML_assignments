```python
#question 4 : most frequent value and indices

import numpy as np

x = np.array([1,2,3,4,5,1,2,1,1,1])
most_freq_x =np.argmax(np.bincount(x))
indicies_x=np.where(x==most_freq_x)[0]
print("most frequent value in x : ",most_freq_x)
print("indicies of most frequent value in x: ",indicies_x)

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
most_freq_y =np.argmax(np.bincount(y))
indicies_y=np.where(y==most_freq_y)[0]
print("most frequent value in y : ",most_freq_y)
print("indicies of most frequent value in y: ",indicies_y)
```

```python
#question 5 : sum of all elements

import numpy as np
gfg = np.matrix('[4, 1, 9; 12, 3, 1; 4, 5, 6]')

sum_all_elements= np.sum(gfg)
sum_row_wise= np.sum(gfg,axis=1)
sum_col_wise= np.sum(gfg,axis=0)

print("Sum of all elements: ",sum_all_elements)
print("sum of all elements row wise: ",sum_row_wise)
print("sum of all elements column wise: ",sum_col_wise)
```

```python
#question 6 : matrix operations

import numpy as np
n_array = np.array([[55, 25, 15],[30, 44, 2],[11, 45, 77]])

sum_of_diagonal= np.trace(n_array)
print("sum_of_diagonal elements:\n",sum_of_diagonal)

eigenvalues,eigenvectors=np.linalg.eig(n_array)
print("eigen values of matrix:\n",eigenvalues)
print("eigen vector of matrix:\n",eigenvectors)

inverse_matrix=np.linalg.inv(n_array)
print("inverse of matrix:\n",inverse_matrix)

determinant= np.linalg.det(n_array)
print("determinant of matrix: ",determinant)
```

```python
#question 7 : mupliplication and covariance

import numpy as np
p1= np.array([[1,2],[2,3]])
q1= np.array([[4,5],[6,7]])

product_1= np.dot(p1,q1)
print("matrix multplication of 1 : \n",product_1)

p1_flatten=p1.flatten()
q1_flatten=q1.flatten()
covariance_1= np.cov(p1_flatten,q1_flatten)[0,1]     #covariance is obtained from [0,1] element of matrix
print("covariance between matrices of 1: ",covariance_1)


p2= np.array([[1, 2], [2, 3], [4, 5]])
q2=np.array([[4, 5, 1], [6, 7, 2]])

product_2= np.dot(p2,q2)
print("matrix multplication of 2 : \n",product_2)

p2_flatten=p2.flatten()
q2_flatten=q2.flatten()
covariance_2= np.cov(p2_flatten,q2_flatten)[0,1]
print("covariance between matrices of 2: ",covariance_2)
```

```python
#question 8 : cartesian product
import numpy as np
x = np.array([[2, 3, 4], [3, 2, 9]])
y = np.array([[1, 5, 0], [5, 10, 3]])

inner_product=np.inner(x,y)
print("inner product: \n",inner_product)

outer_product=np.outer(x,y)
print("outer product: \n",outer_product)

x_flat= x.flatten()
y_flat= y.flatten()
x1,y1= np.meshgrid(x_flat,y_flat)
cartesian_product= np.array([x1.ravel(),y1.ravel()]).T
print("cartesian product: ",cartesian_product)

```

```python
import numpy as np

array = np.array([[1, -2, 3],[-4, 5, -6]])

absolute_array=np.abs(array)
print("element-wise absolute value: \n",absolute_array)

flattened_array= array.flatten()
percentile_flattened= np.percentile(flattened_array,[25,50,75])
percentile_column= np.percentile(array,[25,50,75],axis=0)
percentile_row= np.percentile(array,[25,50,75],axis=1)

print("percentile of flattened array(25,50,75): \n",percentile_flattened)
print("percentile of each column(25,50,75): \n",percentile_column)
print("percentile of each row(25,50,75): \n",percentile_row)

mean_flattened=np.mean(flattened_array)
median_flattened=np.median(flattened_array)
std_flattened=np.std(flattened_array)

mean_column=np.mean(array,axis=0)
median_column=np.median(array,axis=0)
std_column=np.std(array,axis=0)

mean_rows=np.mean(array,axis=1)
median_rows=np.median(array,axis=1)
std_rows=np.std(array,axis=1)

print("Mean ,median and standard deviation of flattened array : ")
print("mean: ",mean_flattened)
print("median: ",median_flattened)
print("standard deviation: ",std_flattened)

print("Mean ,median and standard deviation of of each column : ")
print("mean: ",mean_column)
print("median: ",median_column)
print("standard deviation: ",std_column)

print("Mean ,median and standard deviation of each row : ")
print("mean: ",mean_rows)
print("median: ",median_rows)
print("standard deviation: ",std_rows)


```

```python
import numpy as np

a = np.array([-1.8, -1.6, -0.5, 0.5,1.6, 1.8, 3.0])

floor_val=np.floor(a)
ceiling_val=np.ceil(a)
truncated_val=np.trunc(a)
rounded_val=np.round(a)

print("floor value: \n",floor_val)
print("ceiling value: \n",ceiling_val)
print("truncated value: \n",truncated_val)
print("rounded value: \n",rounded_val)
```
